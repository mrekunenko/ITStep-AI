# Завдання 7
Застосуйте метод градієнтного для розв'язання рівняння f(x)=y_true

```python
def f(x):
  return x**2/3 + torch.sin(torch.pi*x)

y_true = 3
```

```python
import torch
from torch.optim import SGD

x = torch.tensor(-3.5, requires_grad=True)

optimizer = SGD(params=[x], lr=0.01)

for epoch in range(100):

    nn_result = f(x)
    
    error = (y_true - nn_result)**2
    
    error.backward()
    
    optimizer.step()
    
    optimizer.zero_grad()

    print(epoch, error)
```

```python
f(x)
```

```python
import matplotlib.pyplot as plt

xs = torch.linspace(-3, 5, 100)
plt.plot(xs.cpu(), f(xs).cpu(), label='f(x)')
plt.scatter(x.item(), f(x).item(), c='red', label='min')
plt.legend()
```
