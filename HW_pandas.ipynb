# Завдання 6
Перевірте чи правильно розраховано загальну вартість покупки. Виведіть ті рядки де є помилка. Виправте дані в таблиці

```python
# Перевіка чи правильно розраховано загальну вартість покупки
df['calculated_amt'] = df['units'] * df['unit_price']
mask = df['sale_amt'] != df['calculated_amt']
df[mask]

# Корегування даних в таблиці
df['sale_amt'] = df['units'] * df['unit_price']
df.drop(columns=['calculated_amt'], inplace=True)
df
```
